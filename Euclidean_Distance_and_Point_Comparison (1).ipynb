```python
point_A = (3, 4)
point_B = (8, 10)

print("Point A:", point_A)
print("Point B:", point_B)
```

```python
dx = point_B[0] - point_A[0]
dy = point_B[1] - point_A[1]

print("Difference in x:", dx)
print("Difference in y:", dy)
```

```python
import math

distance_manual = math.sqrt(dx**2 + dy**2)

print("Euclidean Distance:", round(distance_manual, 4))
```

```python
def euclidean_distance(p1, p2):
    squared_difference = 0
    for a, b in zip(p1, p2):
        squared_difference += (b - a) ** 2
    return math.sqrt(squared_difference)

check_distance = euclidean_distance(point_A, point_B)
print("Distance from function:", round(check_distance, 4))
```

```python
reference = (5, 5)
candidates = {
    "Point 1": (6, 6),
    "Point 2": (12, 5),
    "Point 3": (4, 7),
    "Point 4": (1, 1)
}

distances = {}
for name, point in candidates.items():
    distances[name] = euclidean_distance(reference, point)

for name, distance in distances.items():
    print(f"{name}: {distance:.4f}")
```

```python
nearest_name = min(distances, key=distances.get)
nearest_distance = distances[nearest_name]

print("Nearest candidate:", nearest_name)
print("Minimum distance:", round(nearest_distance, 4))
```

```python
point_C = (2, 3, 4)
point_D = (5, 7, 9)

distance_3d = euclidean_distance(point_C, point_D)
print("3D Point C:", point_C)
print("3D Point D:", point_D)
print("3D Euclidean Distance:", round(distance_3d, 4))
```

```python
forward = euclidean_distance(point_A, point_B)
backward = euclidean_distance(point_B, point_A)

print("A to B:", round(forward, 6))
print("B to A:", round(backward, 6))

assert round(forward, 10) == round(backward, 10)
print("Symmetry check passed!")
```
